# Incremental loading

```python
checkpoint_location = "abfss://bronze@adlsamazonprime.dfs.core.windows.net/checkpoint"
```

```python
df = spark.readStream.format("cloudFiles") \
            .option("cloudFiles.format", "csv") \
            .option("cloudFiles.schemaLocation", checkpoint_location)\
            .load("abfss://raw@adlsamazonprime.dfs.core.windows.net/")
```

```python
display(df)
```

```python
df.writeStream.option("checkpointLocation", checkpoint_location)\
              .trigger(processingTime = '10 seconds')\
              .start("abfss://bronze@adlsamazonprime.dfs.core.windows.net/AmazonPrime_titles")
```
