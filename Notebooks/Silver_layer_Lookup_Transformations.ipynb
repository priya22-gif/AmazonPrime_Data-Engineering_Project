### Parameters

```python
dbutils.widgets.text("sourcefolder","")
dbutils.widgets.text("targetfolder","")
```

```python
source_folder = dbutils.widgets.get("sourcefolder")
target_folder = dbutils.widgets.get("targetfolder")
```

## Data reading

```python
source_path = "abfss://bronze@adlsamazonprime.dfs.core.windows.net"
sink_path = "abfss://silver@adlsamazonprime.dfs.core.windows.net"
```

```python
df = spark.read.format("csv")\
           .option("header", "true")\
           .option("inferSchema", "true")\
           .load(f"{source_path}/{source_folder}")
```

```python
display(df)
```

## Data writing

```python
df.write.format("delta")\
        .mode("overwrite")\
        .save(f"{sink_path}/{target_folder}")
```
